```python
from cellphi import Cellphi
import cv2
import numpy as np
```

```python
raw_img = cv2.imread("input/36.jpg")
cellphi = Cellphi(scale_factor=1800.0/370.0, enable_gpu=True, width=926, height=695)
resized_img = cellphi.resize(raw_img)
masks = cellphi.pose(resized_img)
output_img, output_data = cellphi.take(resized_img, masks, draw=['outline', 'label', 'max_feret_diameter'])
cv2.imwrite("36_output.jpg", output_img)
np.mean([d['max_feret_diameter'] for d in output_data])
```
